## 2.3 Missingness

In this section, we study missing values across all features.

Missingness is not only a data quality issue.  
In an integrated credit-risk dataset, missing values may also indicate that a customer has no prior record in a related data source.

This section only collects information.

We will look at:

- missing count and missing percentage  
- train-test missingness difference  
- missingness bands  
- missingness by source group  
- top missing features  
- missingness correlation between features  

No imputation, feature dropping, or missing-indicator creation is performed here.

```python
# Create missingness profile for all available columns

import numpy as np
import matplotlib.pyplot as plt
from pathlib import Path

missing_report_dir = Path("reports") / "missingness"
missing_figure_dir = missing_report_dir / "figures"

missing_report_dir.mkdir(parents=True, exist_ok=True)
missing_figure_dir.mkdir(parents=True, exist_ok=True)

all_columns = sorted(set(train_df.columns) | set(test_df.columns) | set(feature_inventory["feature"]))

inventory_cols = [
    "feature",
    "role",
    "source_group",
    "dtype_semantic",
    "feature_family",
    "profile_route"
]

inventory_base = (
    feature_inventory[inventory_cols]
    .drop_duplicates(subset=["feature"])
    .set_index("feature")
)

missingness_profile = pd.DataFrame({"feature": all_columns})

missingness_profile["train_present"] = missingness_profile["feature"].isin(train_df.columns)
missingness_profile["test_present"] = missingness_profile["feature"].isin(test_df.columns)

missingness_profile["missing_count_train"] = missingness_profile["feature"].apply(
    lambda col: train_df[col].isna().sum() if col in train_df.columns else np.nan
)

missingness_profile["missing_pct_train"] = missingness_profile["feature"].apply(
    lambda col: train_df[col].isna().mean() * 100 if col in train_df.columns else np.nan
)

missingness_profile["missing_count_test"] = missingness_profile["feature"].apply(
    lambda col: test_df[col].isna().sum() if col in test_df.columns else np.nan
)

missingness_profile["missing_pct_test"] = missingness_profile["feature"].apply(
    lambda col: test_df[col].isna().mean() * 100 if col in test_df.columns else np.nan
)

missingness_profile["missing_gap_pct"] = (
    missingness_profile["missing_pct_train"] - missingness_profile["missing_pct_test"]
)

missingness_profile["missing_gap_abs"] = missingness_profile["missing_gap_pct"].abs()

missingness_profile["all_null_train"] = missingness_profile["missing_pct_train"].eq(100)
missingness_profile["all_null_test"] = missingness_profile["missing_pct_test"].eq(100)

missingness_profile = missingness_profile.join(inventory_base, on="feature")

missingness_profile
```

```python
# Add missingness bands and basic source hint

def missingness_band(pct):
    if pd.isna(pct):
        return "not_available"
    if pct == 0:
        return "none"
    if pct <= 5:
        return "low"
    if pct <= 20:
        return "moderate"
    if pct <= 50:
        return "high"
    if pct <= 80:
        return "very_high"
    return "extreme"

history_source_groups = [
    "bureau",
    "bureau_balance",
    "previous_application",
    "installments",
    "credit_card",
    "pos_cash"
]

missingness_profile["missingness_band_train"] = missingness_profile["missing_pct_train"].apply(missingness_band)
missingness_profile["missingness_band_test"] = missingness_profile["missing_pct_test"].apply(missingness_band)

missingness_profile["history_source_group"] = missingness_profile["source_group"].isin(history_source_groups)

missingness_profile["missing_rank_train"] = (
    missingness_profile["missing_pct_train"]
    .rank(method="dense", ascending=False)
)

missingness_profile = missingness_profile.sort_values(
    ["missing_pct_train", "missing_gap_abs"],
    ascending=False
)

missingness_profile.head(10)
```

```python
# Create compact missingness summaries

band_order = ["none", "low", "moderate", "high", "very_high", "extreme", "not_available"]

missingness_band_summary = (
    missingness_profile["missingness_band_train"]
    .value_counts(dropna=False)
    .reindex(band_order, fill_value=0)
    .reset_index()
)

missingness_band_summary.columns = ["missingness_band_train", "feature_count"]

source_group_missing_summary = (
    missingness_profile
    .groupby("source_group", dropna=False)
    .agg(
        feature_count=("feature", "count"),
        avg_missing_pct_train=("missing_pct_train", "mean"),
        median_missing_pct_train=("missing_pct_train", "median"),
        max_missing_pct_train=("missing_pct_train", "max"),
        features_with_missing=("missing_pct_train", lambda x: (x > 0).sum()),
        extreme_missing_features=("missingness_band_train", lambda x: (x == "extreme").sum())
    )
    .reset_index()
    .sort_values("avg_missing_pct_train", ascending=False)
)

top_missing_features = (
    missingness_profile
    .loc[missingness_profile["missing_pct_train"].fillna(0) > 0]
    .sort_values("missing_pct_train", ascending=False)
    .head(50)
)

top_missing_gap_features = (
    missingness_profile
    .loc[missingness_profile["missing_gap_abs"].fillna(0) > 0]
    .sort_values("missing_gap_abs", ascending=False)
    .head(50)
)
```

The only safe automatic exclusions are non-feature columns such as ID, target, and accidental index artifacts.

Possible later preprocessing direction:

| Feature Situation | Possible Later Handling |
|---|---|
| No missing values | No imputation needed |
| Low/moderate missing numeric features | Median imputation |
| High-missing application property features | Median imputation + missing indicator |
| Missing external score features | Median imputation + missing indicator |

| Categorical features with missing values | Treat missing as a separate category |
| Binary/status features with missing values | Treat missing as unknown or add missing indicator |

Final note: missingness will be carried forward as evidence. Final imputation, feature dropping, and missing-indicator decisions will be made later after checking feature meaning, target relationship, redundancy, and model validation.

### 2.4.1 Which features are included in numeric profiling?

### 2.4.3 What do the numeric features look like statistically?

Here we calculate basic numeric statistics for each selected feature.

This table is the main numeric evidence table.  
It gives us enough information to think later about imputation, scaling, clipping, transformation, and outlier handling.

```python
# Build numeric univariate profile

numeric_train = train_df[numeric_features]

quantiles_train = numeric_train.quantile(
    [0.01, 0.05, 0.25, 0.50, 0.75, 0.95, 0.99]
).T

quantiles_train.columns = ["p1", "p5", "p25", "p50", "p75", "p95", "p99"]

numeric_univariate = pd.DataFrame({
    "feature": numeric_features,
    "n_total_train": train_df.shape[0],
    "n_non_missing_train": numeric_train.count().values,
    "n_unique_train": numeric_train.nunique(dropna=True).values,
    "mean": numeric_train.mean().values,
    "median": numeric_train.median().values,
    "std": numeric_train.std().values,
    "min": numeric_train.min().values,
    "max": numeric_train.max().values,
    "skew": numeric_train.skew().values,
    "kurtosis": numeric_train.kurtosis().values,
    "zero_pct": numeric_train.eq(0).mean().values * 100,
    "negative_pct": numeric_train.lt(0).mean().values * 100
})

numeric_univariate = numeric_univariate.merge(
    quantiles_train.reset_index().rename(columns={"index": "feature"}),
    on="feature",
    how="left"
)

q1 = quantiles_train["p25"]
q3 = quantiles_train["p75"]
iqr = q3 - q1

lower_fence = q1 - 1.5 * iqr
upper_fence = q3 + 1.5 * iqr

iqr_outlier_pct = (
    numeric_train.lt(lower_fence, axis=1) |
    numeric_train.gt(upper_fence, axis=1)
).sum() / numeric_train.notna().sum() * 100

numeric_univariate["iqr_outlier_pct"] = numeric_univariate["feature"].map(iqr_outlier_pct)

numeric_univariate = numeric_univariate.merge(
    numeric_feature_inventory[
        ["feature", "source_group", "dtype_semantic", "feature_family", "profile_route"]
    ],
    on="feature",
    how="left"
)

numeric_univariate.head()
```

### 2.4.4 How different are train and test numeric values?

Missingness can be similar between train and test, but numeric values may still shift.

Here we collect a small train-test numeric comparison using median and upper percentile values.

```python
# Add simple train-test numeric comparison

common_numeric_features = [
    feature for feature in numeric_features
    if feature in test_df.columns and pd.api.types.is_numeric_dtype(test_df[feature])
]

numeric_test = test_df[common_numeric_features]

test_numeric_summary = pd.DataFrame({
    "feature": common_numeric_features,
    "median_test": numeric_test.median().values,
    "p95_test": numeric_test.quantile(0.95).values
})

numeric_univariate = numeric_univariate.merge(
    test_numeric_summary,
    on="feature",
    how="left"
)

numeric_univariate["median_gap_abs"] = (
    numeric_univariate["median"] - numeric_univariate["median_test"]
).abs()

numeric_univariate["p95_gap_abs"] = (
    numeric_univariate["p95"] - numeric_univariate["p95_test"]
).abs()

numeric_univariate.head()
```

### 2.4.5 Which numeric patterns appear most often?

Now we add simple review bands.

These bands are not final decisions.  
They only help us summarize many numeric features in a readable way.

```python
# Add simple numeric review bands

def skew_band(value):
    if pd.isna(value):
        return "not_available"
    value = abs(value)
    if value < 0.5:
        return "roughly_symmetric"
    if value < 1:
        return "moderate_skew"
    if value < 2:
        return "high_skew"
    return "extreme_skew"

def zero_band(value):
    if pd.isna(value):
        return "not_available"
    if value >= 95:
        return "extreme_zero_heavy"
    if value >= 80:
        return "zero_heavy"
    if value > 0:
        return "some_zero"
    return "no_zero"

def outlier_band(value):
    if pd.isna(value):
        return "not_available"
    if value > 10:
        return "severe_outlier_presence"
    if value > 5:
        return "material_outlier_presence"
    if value > 1:
        return "mild_outlier_presence"
    return "low_outlier_presence"

numeric_univariate["skew_band"] = numeric_univariate["skew"].apply(skew_band)
numeric_univariate["zero_band"] = numeric_univariate["zero_pct"].apply(zero_band)
numeric_univariate["outlier_band"] = numeric_univariate["iqr_outlier_pct"].apply(outlier_band)

numeric_univariate.head()
```

### 2.4.6 What are the main numeric summary views?

Here we summarize skewness, zero values, outlier evidence, and source-group patterns.

```python
# Create compact numeric summaries

skew_band_summary = (
    numeric_univariate["skew_band"]
    .value_counts(dropna=False)
    .reset_index()
)

skew_band_summary.columns = ["skew_band", "feature_count"]

zero_band_summary = (
    numeric_univariate["zero_band"]
    .value_counts(dropna=False)
    .reset_index()
)

zero_band_summary.columns = ["zero_band", "feature_count"]

outlier_band_summary = (
    numeric_univariate["outlier_band"]
    .value_counts(dropna=False)
    .reset_index()
)

outlier_band_summary.columns = ["outlier_band", "feature_count"]

source_numeric_summary = (
    numeric_univariate
    .groupby("source_group", dropna=False)
    .agg(
        feature_count=("feature", "count"),
        avg_abs_skew=("skew", lambda x: x.abs().mean()),
        avg_zero_pct=("zero_pct", "mean"),
        avg_iqr_outlier_pct=("iqr_outlier_pct", "mean"),
        median_missing_ready_count=("n_non_missing_train", "count")
    )
    .reset_index()
    .sort_values("feature_count", ascending=False)
)

display(skew_band_summary)
display(zero_band_summary)
display(outlier_band_summary)
display(source_numeric_summary)
```

## 2.5 Categorical Profile

In this section, we profile categorical features.

The goal is to collect information needed for later encoding decisions.

We are not applying one-hot encoding, ordinal encoding, rare-category grouping, missing filling, or feature dropping here.

We will collect information about:

- category count  
- dominant categories

- missing categories  
- train-test category compatibility  
- unseen categories in test  

### 2.5.1 What categorical features are we profiling?

Before profiling, we first select categorical model features from the manual feature inventory.

```python
# Select categorical features from the manual inventory

import numpy as np
import matplotlib.pyplot as plt
from pathlib import Path

categorical_report_dir = Path("reports") / "categorical_profile"
categorical_figure_dir = categorical_report_dir / "figures"

categorical_report_dir.mkdir(parents=True, exist_ok=True)
categorical_figure_dir.mkdir(parents=True, exist_ok=True)

categorical_feature_inventory = feature_inventory[
    (feature_inventory["dtype_semantic"] == "categorical") &
    (feature_inventory["role"] == "model_feature") &
    (feature_inventory["feature"].isin(train_df.columns))
].copy()

categorical_features = categorical_feature_inventory["feature"].tolist()

categorical_scope_summary = pd.DataFrame({
    "item": [
        "categorical_features_selected",
        "train_rows",
        "test_rows"
    ],
    "value": [
        len(categorical_features),
        train_df.shape[0],
        test_df.shape[0]
    ]
})

display(categorical_scope_summary)

categorical_feature_inventory[
    ["feature", "source_group", "dtype_semantic", "feature_family", "profile_route"]
]
```

### 2.5.2 What is the categorical feature structure?

Here we check how categorical features are distributed by source group and feature family.

```python
# Create categorical grouping summaries

categorical_source_summary = (
    categorical_feature_inventory["source_group"]
    .value_counts(dropna=False)
    .reset_index()
)

categorical_source_summary.columns = ["source_group", "feature_count"]

categorical_family_summary = (
    categorical_feature_inventory["feature_family"]
    .value_counts(dropna=False)
    .reset_index()
)

categorical_family_summary.columns = ["feature_family", "feature_count"]

display(categorical_source_summary)
display(categorical_family_summary)
```

### 2.5.3 What are the cardinality, dominance, rarity, and missing patterns?

This is the main categorical evidence table.

For each categorical feature, we collect unique counts, top categories, rare-category information, and missing percentage.

```python
# Build categorical univariate profile

rare_threshold_count = max(50, int(np.ceil(0.005 * train_df.shape[0])))

categorical_rows = []

for feature in categorical_features:
    train_series = train_df[feature]
    test_series = test_df[feature] if feature in test_df.columns else pd.Series(dtype="object")
    
    train_counts = train_series.value_counts(dropna=True)
    test_counts = test_series.value_counts(dropna=True)
    
    top_category = train_counts.index[0] if len(train_counts) > 0 else np.nan
    top_category_share = train_counts.iloc[0] / train_df.shape[0] * 100 if len(train_counts) > 0 else np.nan
    
    second_category = train_counts.index[1] if len(train_counts) > 1 else np.nan
    second_category_share = train_counts.iloc[1] / train_df.shape[0] * 100 if len(train_counts) > 1 else np.nan
    
    rare_counts = train_counts[train_counts < rare_threshold_count]
    
    train_unique_values = set(train_series.dropna().unique())
    test_unique_values = set(test_series.dropna().unique()) if feature in test_df.columns else set()
    unknown_values_in_test = test_unique_values - train_unique_values
    
    unknown_in_test_share = (
        test_series.isin(unknown_values_in_test).mean() * 100
        if feature in test_df.columns and len(unknown_values_in_test) > 0
        else 0
    )
    
    categorical_rows.append({
        "feature": feature,
        "n_total_train": train_df.shape[0],
        "n_non_missing_train": train_series.notna().sum(),
        "missing_pct_train": train_series.isna().mean() * 100,
        "missing_pct_test": test_series.isna().mean() * 100 if feature in test_df.columns else np.nan,
        "n_unique_train": train_series.nunique(dropna=True),
        "n_unique_test": test_series.nunique(dropna=True) if feature in test_df.columns else np.nan,
        "top_category_train": top_category,
        "top_category_share_train": top_category_share,
        "second_category_train": second_category,
        "second_category_share_train": second_category_share,
        "rare_threshold_count": rare_threshold_count,
        "rare_category_count_train": len(rare_counts),
        "rare_category_share_train": rare_counts.sum() / train_df.shape[0] * 100,
        "unknown_in_test_count": len(unknown_values_in_test),
        "unknown_in_test_share": unknown_in_test_share
    })

categorical_univariate = pd.DataFrame(categorical_rows)

categorical_univariate = categorical_univariate.merge(
    categorical_feature_inventory[
        ["feature", "source_group", "dtype_semantic", "feature_family", "profile_route"]
    ],
    on="feature",
    how="left"
)

categorical_univariate
```

### 2.5.4 Are there unseen categories in test?

Here we add simple information labels for cardinality, dominance, and unseen test categories.

These labels are only for review. They are not final encoding decisions.

```python
# Add categorical review bands

def cardinality_band(n_unique):
    if pd.isna(n_unique):
        return "not_available"
    if n_unique == 2:
        return "binary"
    if 3 <= n_unique <= 10:
        return "low"
    if 11 <= n_unique <= 30:
        return "medium"
    if 31 <= n_unique <= 100:
        return "high"
    if n_unique > 100:
        return "extreme"
    return "single_or_empty"

def dominance_band(top_share):
    if pd.isna(top_share):
        return "not_available"
    if top_share > 99:
        return "near_constant"
    if top_share > 95:
        return "mostly_constant"
    if top_share > 80:
        return "dominant"
    return "mixed"

categorical_univariate["cardinality_band"] = categorical_univariate["n_unique_train"].apply(cardinality_band)
categorical_univariate["dominance_band"] = categorical_univariate["top_category_share_train"].apply(dominance_band)

categorical_univariate["is_binary_like"] = categorical_univariate["n_unique_train"].eq(2)
categorical_univariate["is_near_constant"] = categorical_univariate["top_category_share_train"].gt(99)
categorical_univariate["unseen_test_flag"] = categorical_univariate["unknown_in_test_count"].gt(0)

categorical_univariate.head()
```

### 2.5.5 Which categorical features need manual review later?

These summary tables help us see cardinality, dominance, rarity, missingness, and unseen-test patterns.

No feature decision is made here.

```python
# Create compact categorical summaries

cardinality_summary = (
    categorical_univariate["cardinality_band"]
    .value_counts(dropna=False)
    .reset_index()
)

cardinality_summary.columns = ["cardinality_band", "feature_count"]

dominance_summary = (
    categorical_univariate["dominance_band"]
    .value_counts(dropna=False)
    .reset_index()
)

dominance_summary.columns = ["dominance_band", "feature_count"]

source_categorical_summary = (
    categorical_univariate
    .groupby("source_group", dropna=False)
    .agg(
        feature_count=("feature", "count"),
        avg_unique_train=("n_unique_train", "mean"),
        avg_top_category_share=("top_category_share_train", "mean"),
        features_with_unseen_test=("unseen_test_flag", "sum"),
        avg_missing_pct_train=("missing_pct_train", "mean")
    )
    .reset_index()
    .sort_values("feature_count", ascending=False)
)

high_cardinality_features = (
    categorical_univariate
    .sort_values("n_unique_train", ascending=False)
)

dominant_categorical_features = (
    categorical_univariate
    .sort_values("top_category_share_train", ascending=False)
)

rare_heavy_categorical_features = (
    categorical_univariate
    .sort_values("rare_category_share_train", ascending=False)
)

unseen_test_category_features = (
    categorical_univariate
    .loc[categorical_univariate["unseen_test_flag"]]
    .sort_values("unknown_in_test_share", ascending=False)
)

missing_categorical_features = (
    categorical_univariate
    .loc[categorical_univariate["missing_pct_train"] > 0]
    .sort_values("missing_pct_train", ascending=False)
)

display(cardinality_summary)
display(dominance_summary)
display(source_categorical_summary)

display(high_cardinality_features[
    ["feature", "source_group", "n_unique_train", "n_unique_test", "cardinality_band"]
])

display(dominant_categorical_features[
    ["feature", "source_group", "top_category_train", "top_category_share_train", "dominance_band"]
])

display(rare_heavy_categorical_features[
    ["feature", "source_group", "rare_category_count_train", "rare_category_share_train"]
])

display(unseen_test_category_features[
    ["feature", "source_group", "unknown_in_test_count", "unknown_in_test_share"]
])

display(missing_categorical_features[
    ["feature", "source_group", "missing_pct_train", "missing_pct_test"]
])
```
